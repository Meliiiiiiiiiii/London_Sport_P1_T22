```python
import pyreadstat

_,meta=pyreadstat.read_sav("../ActiveLives_Data/surveydata1617.sav",metadataonly=True)
print("total columns:",len(meta.column_names))

# find borough/LA columns
la_cols=[c for c in meta.column_names if "la" in c.lower() or "borough" in c.lower() or "authority" in c.lower()]
print("borough columns:",la_cols)
# find msoa/lsoa columns
geo_cols=[c for c in meta.column_names if "msoa" in c.lower() or "lsoa" in c.lower()]
print("LSOA columns:",geo_cols)
# find demographic columns (age, disability, nssec etc used in poplprofile)
demo_cols=[c for c in meta.column_names if any(k in c.lower() for k in ["age","gender","sex","disab","nssec","ethnic","imd"])]
print("demographic columns:",demo_cols[:30])  # first 30 more likely
```
