```python
import pandas as pd

PATH_2223 = r"C:\Users\ouame\Downloads\2022-2023\UKDA-9288-spss\spss\spss28\active_lives_survey_nov_22-23_data_year_8_shared_20250103.sav"
SURVEY_YEAR_LABEL_2223 = "2022-23"

cols = ["Reg9", "LA_2023", "wt_final", "MEMS7GR_ALL"]
df_2223, meta_2223 = pyreadstat.read_sav(PATH_2223, usecols=cols, apply_value_formats=True)

london_2223 = df_2223[df_2223["Reg9"] == "London"].copy()
london_2223["borough"] = london_2223["LA_2023"].str.replace(r"^E\d{8}\s+", "", regex=True)

def weighted_pct(group, category):
    mask = group["MEMS7GR_ALL"] == category
    return 100 * group.loc[mask, "wt_final"].sum() / group["wt_final"].sum()

rows = []
for borough, group in london_2223.groupby("borough"):
    rows.append({
        "survey_year": SURVEY_YEAR_LABEL_2223,
        "borough": borough,
        "pct_active": round(weighted_pct(group, "Active"), 1),
        "pct_fairly_active": round(weighted_pct(group, "Fairly Active"), 1),
        "pct_inactive": round(weighted_pct(group, "Inactive"), 1),
        "respondents": len(group),
        "weighted_base": round(group["wt_final"].sum(), 1),
    })

gap_score_2223 = pd.DataFrame(rows).sort_values("borough").reset_index(drop=True)
gap_score_2223.to_csv("active_lives_2022-23_london_gapscore.csv", index=False)

print(f"Boroughs found: {len(gap_score_2223)} (should be 33)")
gap_score_2223
```

```python
MIN_SAMPLE_SIZE = 30

DEMOGRAPHIC_VARS = [
    "Age9", "Gend3", "Eth7", "IMD10", "Disab3", "LondInOut",
    "NSSEC5", "Educ6", "Orient4", "Relig7", "ChildAgeU13", "Maternity_pop",
]

cols2 = ["Reg9", "LA_2023", "wt_final", "MEMS7GR_ALL"] + DEMOGRAPHIC_VARS
df2_2223, meta2_2223 = pyreadstat.read_sav(PATH_2223, usecols=cols2, apply_value_formats=True)

london2_2223 = df2_2223[df2_2223["Reg9"] == "London"].copy()
london2_2223["borough"] = london2_2223["LA_2023"].str.replace(r"^E\d{8}\s+", "", regex=True)

def weighted_pct2(group, category):
    total_wt = group["wt_final"].sum()
    if total_wt == 0:
        return None
    mask = group["MEMS7GR_ALL"] == category
    return round(100 * group.loc[mask, "wt_final"].sum() / total_wt, 1)

def build_row(borough_label, dem_var, dem_category, group):
    return {
        "survey_year": SURVEY_YEAR_LABEL_2223,
        "borough": borough_label,
        "demographic_group": dem_var,
        "category": dem_category,
        "pct_active": weighted_pct2(group, "Active"),
        "pct_fairly_active": weighted_pct2(group, "Fairly Active"),
        "pct_inactive": weighted_pct2(group, "Inactive"),
        "respondents": len(group),
        "weighted_base": round(group["wt_final"].sum(), 1),
        "suppress": len(group) < MIN_SAMPLE_SIZE,
    }

rows2 = []
for dem_var in DEMOGRAPHIC_VARS:
    valid = london2_2223[london2_2223[dem_var].notna()]

    for dem_category, group in valid.groupby(dem_var, observed=True):
        rows2.append(build_row("London", dem_var, dem_category, group))

    for (borough, dem_category), group in valid.groupby(["borough", dem_var], observed=True):
        rows2.append(build_row(borough, dem_var, dem_category, group))

population_profile_2223 = pd.DataFrame(rows2)
population_profile_2223.to_csv("activelives_population_profile_2022-23.csv", index=False)

print("Total rows:", len(population_profile_2223))
print("Suppressed rows:", population_profile_2223["suppress"].sum())
population_profile_2223.head(20)
```
